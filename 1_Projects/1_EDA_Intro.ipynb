```python
import pandas as pd
import matplotlib.pyplot as plt
import datetime as dt
import ast
import seaborn as sns
from datasets import load_dataset
```

# Loading the dataset from hugging face and converting it into Pandas dataframe

```python
dataset=load_dataset("lukebarousse/data_jobs")
df=dataset["train"].to_pandas()
df_work=df.copy()
```

# Data cleanup 

```python
# Converting job_posted_date to datetime object 
df_work["job_posted_date"]=pd.to_datetime(df_work["job_posted_date"])
```

```python
# Converting job_skills to list 
df_work["job_skills"]=df_work.job_skills.apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
type(df_work["job_skills"][1])
```

```python
df_Poland= df_work[df.job_country=="Poland"].reset_index()
df_Poland
```

```python
df_DA_PL=df_Poland[df_Poland.job_title_short=="Data Analyst"]
df_DA_PL
```

```python
df_DS_PL=df_Poland[df_Poland.job_title_short=="Data Scientist"]
df_DS_PL
```

## Job locations for Data Anayst positions in Poland

```python
agg_plot=df_DA_PL["job_location"].value_counts().head(10).to_frame()
sns.set_theme(style="ticks")
sns.barplot(data=agg_plot, y=agg_plot.index, x="count", hue="count", palette="light:b")
plt.legend().remove()
sns.despine()
plt.title("Number of data analyst job postings in Poland by city")
plt.ylabel(" ")
plt.xlabel("Number of jobs")
```

## Job oppurtunities 

```python
dict_column={"job_work_from_home":"Work from home", "job_no_degree_mention" : "Degree mention"}
fig, ax=plt.subplots(1,2, figsize=(6,6))
for i, (column, title) in enumerate(dict_column.items()):
    ax[i].pie(df_DA_PL[column].value_counts(), labels=["Yes", "No"], startangle=90, autopct="%1.1f%%")
    ax[i].set_title(title) 
#fig.suptitle("Benefit analysis of the data analyst jobs in Poland")
fig.tight_layout()
```

## Companies to explore

```python
agg_plot=df_DA_PL["company_name"].value_counts().head(10).to_frame()
sns.set_theme(style="ticks")
sns.barplot(data=agg_plot, y=agg_plot.index, x="count", hue="count", palette="light:b")
plt.legend().remove()
sns.despine()
plt.title("Counts of companies for data analyst jobs in Poland")
plt.ylabel(" ")
plt.xlabel("Number of jobs")
```
