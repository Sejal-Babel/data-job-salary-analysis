## What are the most demanded skills for the 3 most popular data roles?

# Methodology 
1) Clean up the skill column
2) Filter by job_title_short
3) Calculate skill percentage
4) Plot the results 

```python
#importing libraries 
import pandas as pd
import matplotlib.pyplot as plt
import datetime as dt
import ast
import seaborn as sns
from datasets import load_dataset


#Loading the data 
dataset=load_dataset("lukebarousse/data_jobs")
df=dataset["train"].to_pandas()
df_work=df.copy()


#Data cleanup 
# Converting job_posted_date to datetime object 
df_work["job_posted_date"]=pd.to_datetime(df_work["job_posted_date"])
# Converting job_skills to list 
df_work["job_skills"]=df_work.job_skills.apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

# Filter data from Poland 

```python
df_Poland= df_work[df.job_country=="Poland"].reset_index()
df_Poland.drop(columns="index", inplace=True)
df_Poland
```

```python
df_skills=df_Poland.explode("job_skills")
df_skills
```

```python
df_skills_count=df_skills.groupby(["job_skills", "job_title_short"]).size().reset_index(name="Count")
df_skills_count.sort_values(by="Count", ascending=False, inplace=True)
df_skills_count["job_skills"]=df_skills_count.job_skills.apply(lambda x: x.title() if pd.notna(x) else x)
df_skills_count
```

```python
df_skills[(df_skills.job_title_short=="Data Engineer") & (df_skills.job_skills=="airflow")]
```

```python
unique_titles=df_skills_count.job_title_short.unique().tolist()
unique_titles
```

```python
job_titles=unique_titles[0:3]
job_titles
```

```python
fig, ax=plt.subplots(len(job_titles), 1, figsize=(7, 8))
for i, title in enumerate(job_titles):
    to_plot=df_skills_count[df_skills_count.job_title_short==title].sort_values("Count", ascending=False).head()
    to_plot.plot(kind="barh", x="job_skills", y="Count", ax=ax[i], title=title)
    ax[i].invert_yaxis()
    ax[i].legend().remove()
    ax[i].set_xlabel("Skill count")
    ax[i].set_ylabel(" ")
plt.suptitle("Popular required skills by job title")
plt.tight_layout()
plt.show()
```

```python
Job_count=df_Poland["job_title_short"].value_counts().reset_index(name="Jobs_Total")
Job_count
```

```python
df_skills_count
```

```python
df_skills_percent=pd.merge(df_skills_count, Job_count, on="job_title_short", how="left")
df_skills_percent["skill_percent"]=(df_skills_percent["Count"]/df_skills_percent["Jobs_Total"])*100
df_skills_percent
```

```python
fig, ax=plt.subplots(len(job_titles), 1, figsize=(7, 8))
sns.set_theme(style="ticks")
for i, title in enumerate(job_titles):
    to_plot=df_skills_percent[df_skills_percent.job_title_short==title].sort_values("skill_percent", ascending=False).head()
    sns.barplot(data=to_plot, x="skill_percent", y="job_skills", ax=ax[i], hue="skill_percent", palette="light:b")
    ax[i].set_title(title)
    ax[i].legend().remove()
    ax[i].set_xlabel(" ")
    ax[i].set_ylabel(" ")
    ax[i].set_xlim(0,70)

    for n, value in enumerate(to_plot["skill_percent"]):
        ax[i].text(x=value+1, y=n, s=f"{round(value,1)}%", va="center")
    if i != len(job_titles)-1:
        ax[i].set_xticks([])
        

plt.suptitle("Likelihood of skill requirement in job postings in Poland")
plt.tight_layout()
plt.show()
```
